```python
text_columns = ["customer_name", "gender", "city", "subscription",
                "discount_applied", "item_purchased", "category",
                "payment_method"]

for columns in text_columns:
    df[columns] = df[columns].str.strip()
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins = [17,25,35,45,60],
    labels=["18-25","26-35","36-45","46-60"]
)
```

to create age group in dataframe

```python
df['purchase_month'] = df["purchase_date"].dt.month_name()
```

to create a month columns in dataframe 

```python
df["purchase_day"] = df["purchase_date"].dt.day_name()
```

to create a day column in dataframe

```python
df["purchase_year"] = df["purchase_date"].dt.year
```

to create year column in dataframe

```python
df.columns
```

```python
df["discount_flag"] = df["discount_applied"].map({
    "Yes":1,
    "No":0
})
```

to create a discount_flag column in dataframe

```python
df.head()
```

```python
df["revenue_segment"] = pd.cut(
    df["purchase_amount"],
    bins = [0,10000,30000,50000,70000],
    labels = ["low","medium","high","premium"]
)
```
